#  A/B Testing — Discount Impact on Revenue & Margin
# Does giving a higher discount actually cause a real difference in revenue, margin, and quantity sold — or does it just look different by chance?

```python
from scipy import stats

# Split into A/B groups based on discount tier
group_a = beauty[beauty['discount_percent'] <= 10]   # Low discount (control)
group_b = beauty[beauty['discount_percent'] > 20]    # High discount (treatment)

print('Group A (<=10% discount) size:', len(group_a))
print('Group B (>20% discount) size:', len(group_b))
```

```python
results = []

for metric in ['net_revenue', 'gross_margin_percent', 'qty_sold']:
    a = group_a[metric]
    b = group_b[metric]
    t_stat, p_val = stats.ttest_ind(a, b, equal_var=False)
    results.append({
        'metric': metric,
        'mean_group_a': round(a.mean(), 2),
        'mean_group_b': round(b.mean(), 2),
        't_stat': round(t_stat, 3),
        'p_value': round(p_val, 5),
        'significant (p<0.05)': p_val < 0.05
    })

ab_results = pd.DataFrame(results)
ab_results
```
